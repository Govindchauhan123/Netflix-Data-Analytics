```python
# Duplicated value
df.duplicated().sum()
```

```python
# Remove duplicates value from data set

df.drop_duplicates(inplace=True)
```

```python
# Dirty value:- null,N/A, blank,space

df.replace(['N/A','NULL','',' '],np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
# Extra Spaces Remove
df = df.apply(
    lambda x: x.str.strip() if x.dtype in ['str', 'object', 'string'] else x
)
```

```python
df = df.map(lambda x: x.title() if isinstance(x, str) else x)
```

```python
df['Country'].value_counts()
```

```python
# Numeric Columns
numeric_cols={

    'Age','Watch_Hours','Monthly_Fee','Movies_Watched','Episodes_Watched','Avg_Rating','Monthly_Sessions','Download_Count'
}
```

```python
for col in numeric_cols:
    df[col]=pd.to_numeric(df[col],errors='coerce')
```

```python
df.describe()
```

```python
df = df[(df['Age'] >= 18) & (df['Age'] <= 100)] 
```

```python
df.describe()
```

```python
#3. Negative values remove karna
# Agar Watch_Hours aur Monthly_Fee dono negative nahi hone chahiye 

df = df[(df['Watch_Hours'] >= 0) & (df['Monthly_Fee'] >= 0)]
```

```python
df['Join_Date'] = pd.to_datetime(
    df['Join_Date'], errors='coerce'
).dt.date

df['Last_Watch_Date'] = pd.to_datetime(
    df['Last_Watch_Date'], errors='coerce'
).dt.date
```

```python
df[['Join_Date','Last_Watch_Date']].isnull().sum()
```

```python
# Missing values handle
df['Age'].fillna(
    df['Age'].mean(), inplace=True
)

df['Watch_Hours'].fillna(
    df['Watch_Hours'].mean(), inplace=True
)



df['Monthly_Fee'].fillna(
    df['Monthly_Fee'].mean(), inplace=True
)
```

```python
df['Payment_Method']=df['Payment_Method'].fillna('Unknown')

df['Device']=df['Device'].fillna('Unknown')

df['Genre']=df['Genre'].fillna('Unknown')
```

```python
df
```

```python
# Feature Engineering
# Customer Revenue

df["Customer_Revenue"]=(df['Monthly_Fee']* df['Monthly_Sessions'])
```

```python
df['Engagement_Score']=(df['Watch_Hours']+df['Monthly_Sessions']+df['Download_Count'])
```

```python
# Watch category
df["Watch_category"] = pd.cut(
    df['Watch_Hours'],
    bins=[-np.inf, 20, 50, 100, np.inf],
    labels=['Low', 'Medium', 'High', 'Very High']
)
```

```python
df['Country'] = df['Country'].fillna('Unknown')
```

```python
# Customer ages group


bins=[17,25,35,50,100]
label=['18-25','26-35','36-50','50+']

df["Age_Group"]=pd.cut(df['Age'],bins=bins,labels=label)
```

```python
# Account Active Flag

df['Active_Flag']=np.where(df["Account_Status"]=="Active", 1,0)
```
